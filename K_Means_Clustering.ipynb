```python
# Range of k values to try
k_values = range(1, 11)
inertia = []

for k in k_values:
    kmeans = KMeans(n_clusters=k, max_iter=300, n_init=20, random_state=42)
    kmeans.fit(df.drop('Private', axis=1))
    inertia.append(kmeans.inertia_)

# Plot the elbow curve
plt.figure(figsize=(10, 6))
plt.plot(k_values, inertia, 'bo-')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method For Optimal k')
plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in k_values[1:]:
    kmeans = KMeans(n_clusters=k, max_iter=300, n_init=20, random_state=42)
    kmeans.fit(df.drop('Private', axis=1))
    score = silhouette_score(df.drop('Private', axis=1), kmeans.labels_)
    silhouette_scores.append(score)

# Plot the silhouette scores
plt.figure(figsize=(10, 6))
plt.plot(k_values[1:], silhouette_scores, 'bo-')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score For Optimal k')
plt.show()
```

```python
optimal_k = k_values[np.argmax(silhouette_scores) + 1]
print(f"The optimal number of clusters is {optimal_k}")

```

```python
kmeans = KMeans(n_clusters=optimal_k, max_iter=300, n_init=20, random_state=42)
kmeans.fit(df.drop('Private', axis=1))
clus_cent = kmeans.cluster_centers_
```
